```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
import warnings

# Suppress warnings for a clean output
warnings.filterwarnings("ignore")

print("Loading the Smart AI model (TinyLlama)... This will take a minute or two.")

# 1. Model Loading (Swapped to a factual, instruction-tuned model)
model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print("\nChatbot: Hello! I am your AI assistant. How can I help you today?")

# We use a list to store the conversation cleanly.
# The "system" prompt tells the AI how to behave before you even speak to it.
chat_history = [
    {"role": "system", "content": "You are a smart, factual, and helpful AI assistant."}
]

# 4. Continuous Conversation
while True:
    # 2. User Input Handling
    user_input = input("User: ")

    # 5. Exit Condition
    if user_input.strip().lower() in ['exit', 'quit']:
        print("Chatbot: Goodbye! Have a great day.")
        break

    # Add the user's new message to the history
    chat_history.append({"role": "user", "content": user_input})

    # Apply the modern "chat template" to format the history exactly how the AI expects it
    prompt = tokenizer.apply_chat_template(chat_history, tokenize=False, add_generation_prompt=True)

    # Translate the formatted text into numbers for the AI
    inputs = tokenizer(prompt, return_tensors="pt")

    # 3. Response Generation
    outputs = model.generate(
        **inputs,
        max_new_tokens=150,  # Limits the response length so it doesn't ramble
        do_sample=True,
        temperature=0.7,     # Keeps the answers focused but natural
        top_k=50,
        top_p=0.95
    )

    # The AI returns the whole conversation. We slice it to get ONLY the brand new words it generated.
    new_tokens = outputs[0][inputs['input_ids'].shape[-1]:]
    response = tokenizer.decode(new_tokens, skip_special_tokens=True)

    print(f"Chatbot: {response}")

    # Add the AI's response to the history so it remembers what it just said for the next loop
    chat_history.append({"role": "assistant", "content": response})
```
